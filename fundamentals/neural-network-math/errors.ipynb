```python
import numpy as np
import matplotlib.pyplot as plt

# Example error data (replace this with your actual error values)
epochs = np.arange(1, 101)  # Example epochs from 1 to 100
errors = np.random.uniform(low=0.1, high=1.0, size=100)  # Example random errors

# Create a line plot of errors over epochs
plt.plot(epochs, errors, marker='o')
plt.xlabel('Epochs')
plt.ylabel('Error')
plt.title('Error Progression')
plt.grid(True)
plt.show()
```

# Calculating Errors


```python
import numpy as np 

# Example predicted and actual values
predicted = np.array([2.5, 3.0, 4.2, 5.1])
actual = np.array([2.0, 2.8, 3.8, 5.2])

# Calculate Errors
mean_absolute_error = np.mean(np.abs(predicted - actual))
mean_squared_error = np.mean((predicted - actual) ** 2)
root_mean_squared_error = np.sqrt(mean_squared_error)

# Plotting the errors
error_labels = ['MAE', 'MSE', 'RMSE']
error_values = [mean_absolute_error, mean_squared_error, root_mean_squared_error]

plt.bar(error_labels, error_values, color='red')
plt.xlabel('Error Metric')
plt.ylabel('Error Value')
plt.title('Error Metrics Comparison')
plt.show()
```
