```python
import numpy as np

house_sizes = np.array([1000, 1500, 2000, 2500, 3000])

house_prices = np.array([200000, 300000, 400000, 500000, 600000])

weight = 0.1
bias = 1000

# Predicting house prices using the linear model
predicted_prices = weight * house_sizes + bias

for size, actual_price, predicted_prices in zip(house_sizes, house_prices, predicted_prices):
    print(f'House size: {size} sqft | Actual Price: ${actual_price} | Predicted Price: ${predicted_prices:.2f}')
```
