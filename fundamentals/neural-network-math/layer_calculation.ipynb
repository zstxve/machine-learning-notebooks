```python
import numpy as np

# Define the input features 
input_features = np.array([0.5, 0.3, 0.7])

# Define the weights and biases for the hidden layer
hidden_weights = np.array([[0.2, 0.4, -0.5],
                          [0.1, -0.3, 0.2]])
hidden_biases = np.array([0.1, 0.2])

# Define the weights and biases for the output layer
output_weights = np.array([0.3, -0.1])
output_bias = np.array([0.05])

# Calculate input to hidden layer
hidden_layer_input = np.dot(hidden_weights, input_features) + hidden_biases
hidden_layer_output = 1 / (1 + np.exp(-hidden_layer_input))

# Calculate input to output layer
output_layer_input = np.dot(output_weights, hidden_layer_output) + output_bias
output_layer_output = 1 / (1 + np.exp(-output_layer_input))

print('Input to Hidden Layer:', hidden_layer_input)
print('Hidden Layer Output:', hidden_layer_output)
print('Input to Output Layer:', output_layer_input)
print('Output of the Network:', output_layer_output)
```
