```python
import numpy as np

# Predicted probabilities 
predicted_probabilities = np.array([0.2, 0.8]) 

# Actual probabilities ( ground truth ) 
actual_probabilities = np.array([0, 1]) 

# Calculate cross entropy
cross_entropy = -np.sum(actual_probabilities * np.log(predicted_probabilities))
print('Cross Entropy', cross_entropy)
```
