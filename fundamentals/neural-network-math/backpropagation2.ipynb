```python
import numpy as np 

# Sigmoid function 
def sigmoid(x):
    return 1 / (1 + np.exp(x))

# Given data
inputs = np.array([[0.7, 0.5], [0.9, 0.3], [0.4, 0.6]])
targets = np.array([[1], [1], [0]])

# Initialize random weights and bias
np.random.seed(42)
weights = np.random.rand(2, 1)
bias = np.random.rand(1)

# Hyperparameters
learn_rate = 0.5
epochs = 500

# Training loop
for epoch in range(epochs):
    # Forward pass
    hidden_layer_input = np.dot(inputs, weights) + bias
    hidden_layer_output = sigmoid(hidden_layer_input)
    
    # Calculate the networks error
    error = targets - hidden_layer_output
    
    # Backpropagation
    output_grad = error * hidden_layer_output * (1 - hidden_layer_output)
    weights += learn_rate * np.dot(inputs.T, output_grad)
    bias += learn_rate * np.sum(output_grad)
    
    # Print loss
    if epoch % 100 == 0:
        loss = np.mean(error ** 2)
        print(f'Epoch {epoch}, Loss: {loss:.4f}')

print('Final Weights:', weights)
print('Final Bias:', bias)
```
