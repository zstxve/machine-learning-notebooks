### Linear Regression in scikit-learn

```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np
```

### Load the data

```python
bmi_life_data = pd.read_csv("LinearRegression/bmi_and_life_expectancy.csv")
bmi_life_data.head(5)
```

```python
import matplotlib.pyplot as plt

x = np.array(bmi_life_data[["BMI"]])
y = np.array(bmi_life_data["Life expectancy"])

def draw_data(x,y):
    for i in range(len(x)):
        plt.scatter(x[i], y[i], color="red", edgecolors="k")
    plt.xlabel("BMI")
    plt.ylabel("Life expectancy")
    
def display(m, b, color="g"):
    r = np.arange(min(x), max(x), 0.1)
    plt.plot(r, m*r+b, color)
```

```python
# Plotting the data
draw_data(x, y)
plt.show()
```

### Build Linear Regression model

```python
model = LinearRegression()
model.fit(x, y)
```

```python
laos_life_exp = model.predict([[21.07931]])
```
