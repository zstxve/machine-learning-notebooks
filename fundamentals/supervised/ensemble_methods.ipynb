# Ada Boost in SKlearn

```python
from sklearn.ensemble import AdaBoostClassifier, RandomForestClassifier, BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read dataset 
df = pd.read_table('SMSSpamCollection.txt', 
                   sep='\t',
                   header=None,
                   names=['label', 'sms_message'])

# Fix response value 
df['label'] = df.label.map({'ham': 0, 'spam':1})

# Split into training and test data
X_train, X_test, y_train, y_test = train_test_split(df['sms_message'],
                                                     df['label'],
                                                     random_state=1)

# Instantiate CountVectorizer
count_vector = CountVectorizer()

# Fit training data 
training_data = count_vector.fit_transform(X_train)

# Transform testing data
testing_data = count_vector.transform(X_test)

# Instantiate Model 
naive_bayes = MultinomialNB()

# Fit model 
naive_bayes.fit(training_data, y_train)

# Predict
predictions = naive_bayes.predict(testing_data)

# Score 
print('Accuracy Score:', format(accuracy_score(y_test, predictions)))
print('Precision Score:', format(precision_score(y_test, predictions)))
```

#### Predict on testing_data

```python
# Predict using BaggingClassifier
bag_preds = b_clf.predict(testing_data)

# Predict using RandomForestClassifier
rf_preds = r_clf.predict(testing_data)

# Predict using AdaBoostClassifier
ada_preds = ada_clf.predict(testing_data)
```

#### Compare Models and predictions

```python
def print_metrics(y_true, preds, model_name=None):
    if model_name == None:
        print('Accuracy Score:', format(accuracy_score(y_true, preds)))
        print('Precision Score:', format(precision_score(y_true, preds)))
        print('Recall Score:', format(recall_score(y_true, preds)))
        print('F1 Score:', format(f1_score(y_true, preds)))
    else:
        print('Accuracy Score for' + " " + model_name + ": " + format(accuracy_score(y_true, preds)))
        print('Precision Score'+ " " + model_name + ": " + format(precision_score(y_true, preds)))
        print('Recall Score' + " " + model_name + ": " + format(recall_score(y_true, preds)))
        print('F1 Score' + " " + model_name + ": " + format(f1_score(y_true, preds)))
        print('\n\n')
```

#### Print Scores

```python
# Bagging Scores
print_metrics(y_test, bag_preds, 'bagging')

# Random Forest 
print_metrics(y_test, rf_preds, 'random forest')

# AdaBoost 
print_metrics(y_test, ada_preds, 'ada boost')

# Naive Bayes
print_metrics(y_test, predictions, 'naive bayes')
```
